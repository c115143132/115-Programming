# 介紹

本程式使用 pyautogui 來完成一個 Blitz 的遊戲機器人。

```python
import pyautogui as pyg
pyg.sleep(2)
basex=50
basey=652
gap=47
```

```python
for row in range(8):
    for col in range(8):
        x = basex + col * gap
        y = basey + row * gap
        color = pyg.pixel(x, y)
        print(row, col, x, y, color)
```

```python
def get_gem_position(r, c):
    x = basex + c * gap
    y = basey + r * gap
    return (x, y)
```

```python
def get_gem_color(r, c, im):
    x, y = get_gem_position(r, c)
    return im.getpixel((x, y))
```

```python
def can_move_left(r, c, im):
    if not (0 <= r < 8 and 1 <= c < 8):
        return False

    left_color = get_gem_color(r, c - 1, im)
    current_color = get_gem_color(r, c, im)

    def color_after_swap(row, col):
        if (row, col) == (r, c):
            return left_color
        if (row, col) == (r, c - 1):
            return current_color
        return get_gem_color(row, col, im)

    def forms_match(row, col):
        color = color_after_swap(row, col)
        for dr, dc in ((0, 1), (1, 0)):
            count = 1
            for direction in (-1, 1):
                nr = row + direction * dr
                nc = col + direction * dc
                while (0 <= nr < 8 and 0 <= nc < 8
                       and color_after_swap(nr, nc) == color):
                    count += 1
                    nr += direction * dr
                    nc += direction * dc
            if count >= 3:
                return True
        return False

    return forms_match(r, c) or forms_match(r, c - 1)
```

```python
def _can_move(r, c, im, dr, dc):
    nr, nc = r + dr, c + dc
    if not (0 <= r < 8 and 0 <= c < 8
            and 0 <= nr < 8 and 0 <= nc < 8):
        return False

    current_color = get_gem_color(r, c, im)
    neighbor_color = get_gem_color(nr, nc, im)

    def color_after_swap(row, col):
        if (row, col) == (r, c):
            return neighbor_color
        if (row, col) == (nr, nc):
            return current_color
        return get_gem_color(row, col, im)

    def forms_match(row, col):
        color = color_after_swap(row, col)
        for dr_match, dc_match in ((0, 1), (1, 0)):
            count = 1
            for direction in (-1, 1):
                mr = row + direction * dr_match
                mc = col + direction * dc_match
                while (0 <= mr < 8 and 0 <= mc < 8
                       and color_after_swap(mr, mc) == color):
                    count += 1
                    mr += direction * dr_match
                    mc += direction * dc_match
            if count >= 3:
                return True
        return False

    return forms_match(r, c) or forms_match(nr, nc)

def can_move_right(r, c, im):
    return _can_move(r, c, im, 0, 1)

def can_move_down(r, c, im):
    return _can_move(r, c, im, 1, 0)

def can_move_up(r, c, im):
    return _can_move(r, c, im, -1, 0)
```

```python
import time

start_time = time.time()
im = pyg.screenshot()
while True:
    if time.time() - start_time >= 60:
        break

    for r in range(8):
        for c in range(8):
            if time.time() - start_time >= 60:
                break

            direction = None
            if can_move_left(r, c, im):
                direction = (r, c - 1)
            elif can_move_right(r, c, im):
                direction = (r, c + 1)
            elif can_move_down(r, c, im):
                direction = (r + 1, c)
            elif can_move_up(r, c, im):
                direction = (r - 1, c)

            if direction is not None:
                pyg.click(*get_gem_position(r, c))
                pyg.click(*get_gem_position(*direction))

        if time.time() - start_time >= 60:
            break

    im = pyg.screenshot()
```
